# Simulation physique — polarisation du sunglint

1. DoLP de Fresnel vs angle d'incidence (angle de Brewster)
2. Distribution de Cox-Munk pour differentes vitesses de vent
3. Carte 2D du DoLP dans l'espace (theta_v, delta_phi)

```python
import numpy as np
import matplotlib.pyplot as plt
from polcal.modeling.geometry import scattering_angle, incidence_angle, facet_tilt_angle, fresnel_coefficients, brewster_angle
from polcal.modeling.glitter import cox_munk_variance, cox_munk_slope_pdf, dolp_fresnel, glint_stokes
from polcal.visualization.plots import plot_dolp_vs_incidence, plot_cox_munk, plot_glint_dolp_map
%matplotlib inline
plt.rcParams['figure.dpi'] = 120
```

## 1. Fresnel et angle de Brewster
theta_B = arctan(n) ~ 53.2 deg  =>  Theta ~ 106 deg

```python
theta_B = brewster_angle(1.34)
print(f'Brewster: {np.degrees(theta_B):.2f} deg')
print(f'Scattering angle at Brewster: {np.degrees(2*theta_B):.2f} deg')
fig = plot_dolp_vs_incidence()
plt.show()
```

## 2. Distribution de Cox-Munk

```python
for W in [2, 5, 8, 12]:
    print(f'W={W} m/s  sigma2={cox_munk_variance(W):.4f}')
fig = plot_cox_munk()
plt.show()
```

## 3. Carte DoLP — theta_s=40deg, W=5m/s
Convention: delta_phi=0 est la direction speculaire.

```python
fig = plot_glint_dolp_map(theta_s_deg=40, wind_speed=5.0)
plt.show()
```

## 4. Sensibilite au vent et a theta_s (plan principal, delta_phi=0)

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
for ax, ts in zip(axes, [20, 40, 60]):
    tv = np.linspace(0, np.radians(70), 200)
    dp = np.zeros_like(tv)
    S = glint_stokes(np.full_like(tv, np.radians(ts)), tv, dp, wind_speed=5.0)
    I, Q = S[...,0], S[...,1]
    dolp = np.where(I>0, np.abs(Q)/I, np.nan)
    ax.plot(np.degrees(tv), dolp, color='darkorange')
    ax.set_title(f'theta_s={ts}deg')
    ax.set_xlabel('theta_v (deg)')
    ax.set_ylabel('DoLP')
    ax.set_ylim(0, 1.05)
fig.tight_layout()
plt.show()
```
